```python
import catboost
boosting_model = catboost.CatBoostClassifier(cat_features=cat_indexes, eval_metric='AUC')


boosting_model.fit(X_train, y_train)

y_train_predicted = boosting_model.predict_proba(X_train)[:, 1]
y_test_predicted = boosting_model.predict_proba(X_test)[:, 1]
```

2. Подбор гиперпараметров

```python
best_lr, best_trees_count, best_test_roc_auc, best_train_roc_auc, best_model = -1, -1, -1, -1, None
for lr in [0.001, 0.01, 0.1, 1]:
  for trees_count in [200, 400, 600, 800, 1000, 1200]:
    boosting_model = catboost.CatBoostClassifier(cat_features=cat_indexes, eval_metric='AUC', learning_rate = lr, iterations = trees_count)
    boosting_model.fit(X_train, y_train)

    y_train_predicted = boosting_model.predict_proba(X_train)[:, 1]
    y_test_predicted = boosting_model.predict_proba(X_test)[:, 1]

    train_auc = roc_auc_score(y_train, y_train_predicted)
    test_auc = roc_auc_score(y_test, y_test_predicted)

    if test_auc > best_test_roc_auc:
      best_test_roc_auc = test_auc
      best_lr = lr
      best_trees_count = trees_count
      best_train_roc_auc = train_auc
      best_model = boosting_model
```
