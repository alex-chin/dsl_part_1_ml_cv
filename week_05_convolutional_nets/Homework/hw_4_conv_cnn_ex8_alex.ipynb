### Задание 8. Реализуйте LeNet

Если мы сделаем параметры сверток обучаемыми, то можем добиться хороших результатов для задач компьютерного зрения. Реализуйте архитектуру LeNet, предложенную еще в 1998 году!
На этот раз используйте модульную структуру (без помощи класса Sequential).

Наша нейронная сеть будет состоять из
* Свёртки 3x3 (1 карта на входе, 6 на выходе) с активацией ReLU;
* MaxPooling-а 2x2;
* Свёртки 3x3 (6 карт на входе, 16 на выходе) с активацией ReLU;
* MaxPooling-а 2x2;
* Уплощения (nn.Flatten);
* Полносвязного слоя со 120 нейронами и активацией ReLU;
* Полносвязного слоя с 84 нейронами и активацией ReLU;
* Выходного слоя из 10 нейронов.




```python
import numpy as np
import matplotlib.pyplot as plt
import torch.nn as nn
import torch
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class LeNet(nn.Module):
    def __init__(self):
        super(LeNet, self).__init__()
        # Первый свёрточный слой: 1 входной канал, 6 выходных каналов, ядро 3x3
        self.conv1 = nn.Conv2d(1, 6, kernel_size=3)
        # Второй свёрточный слой: 6 входных каналов, 16 выходных каналов, ядро 3x3
        self.conv2 = nn.Conv2d(6, 16, kernel_size=3)
        # Уплощение (Flatten)
        self.flatten = nn.Flatten()
        # Полносвязные слои
        self.fc1 = nn.Linear(16 * 5 * 5, 120)  # 5x5 - размер изображения после пулинга
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84, 10)

    def forward(self, x):
        # Применяем первый свёрточный слой и активацию ReLU
        x = F.relu(self.conv1(x))
        # Применяем MaxPooling 2x2
        x = F.max_pool2d(x, 2)
        # Применяем второй свёрточный слой и активацию ReLU
        x = F.relu(self.conv2(x))
        # Применяем MaxPooling 2x2
        x = F.max_pool2d(x, 2)
        # Уплощение (Flatten)
        x = self.flatten(x)
        # Применяем первый полносвязный слой и активацию ReLU
        x = F.relu(self.fc1(x))
        # Применяем второй полносвязный слой и активацию ReLU
        x = F.relu(self.fc2(x))
        # Выходной слой
        x = self.fc3(x)
        return x

# Создаем экземпляр сети
net = LeNet()
print(net)
```
