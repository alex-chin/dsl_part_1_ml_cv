```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Определение модели
model = nn.Sequential(
    nn.Linear(4, 8),
    nn.ReLU(),
    nn.Dropout(p=0.5),
    nn.Linear(8, 1)
)
```

```python
print(model)
```

```python

# Инициализация входных данных и целевых значений
x = torch.randn(1, 4)  # Входные данные
y_true = torch.tensor([[1.0]])  # Истинное значение

# Определение функции потерь и оптимизатора
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)
```

```python

# Прямой проход
y_pred = model(x)

# Вычисление функции потерь
loss = criterion(y_pred, y_true)

# Обратное распространение ошибки
loss.backward()

# Вывод градиентов до обновления весов
print("Градиенты до обновления весов:")
for name, param in model.named_parameters():
    if param.requires_grad:
        print(f"{name}: {param.grad}")
```

```python

# Обновление весов
optimizer.step()

# Вывод весов после обновления
print("\nВеса после обновления:")
for name, param in model.named_parameters():
    if param.requires_grad:
        print(f"{name}: {param.data}")

# Обнуление градиентов
optimizer.zero_grad()
```
